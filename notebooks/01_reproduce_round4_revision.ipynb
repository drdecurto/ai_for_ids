This analysis-only notebook validates the persisted result files and regenerates all final-revision tables. It performs no model fitting, requires no API key, and independently reconstructs the completed second Axis-2 source from its released raw generations.


```python
from pathlib import Path
import json
import subprocess
import sys

ROOT = Path.cwd().resolve()
assert (ROOT / "scripts" / "recompute_round4_revision.py").is_file(), (
    "Run this notebook from the repository root"
)
print(ROOT)
```

```python
subprocess.run(
    [sys.executable, "scripts/recompute_round4_revision.py"],
    cwd=ROOT,
    check=True,
)
```
